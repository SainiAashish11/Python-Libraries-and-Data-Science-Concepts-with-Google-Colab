# Annotations
* ## to place th text over the scatter plot using ***'plt.text()'*** function

```python
# 'batter.csv' dataset having 'name', 'runs' , 'avg' , 'strike_rate'

df = pd.read_csv('/content/drive/MyDrive/CSV_files/Matplotlib Datasets/batter.csv')
df
```

```python
# Now take out the random 25 players from the data
sample_df = df.sample(25)
sample_df
```

# Subplots -> plt.subplots()
- multiple graphs at the same time
- when dealing with multiple column data

```python
# initialising 'batters' dataset
batters = pd.read_csv('/content/drive/MyDrive/CSV_files/Matplotlib Datasets/batter.csv')
```

```python
# a different way to plot graphs
fig , ax = plt.subplots(figsize = (12 , 7))      # 'fig' -> figure and 'ax' -> axes are two objects of subplots

ax.scatter(batters['avg'] , batters['strike_rate'] , color = 'red' , marker = '*')     # ax.scatter()    -> to plot scatter plot using axes
ax.set_title('Batsman Info')                                                           # ax.set_title()  -> to set the 'title' of the graph
ax.set_xlabel('Avg')                                                                   # ax.set_xlabel   -> to set x-axis label
ax.set_ylabel('Strike_rate')                                                           # ax.set_ylabel   -> to set y-axis label

fig.show()              # it is like plt.show()
```

```python
# multiple graphs technique -> plt.subplots(nrows = integer , ncols = integer)

fig , ax = plt.subplots(nrows = 2 , ncols = 1 , sharex = True , figsize = (10 , 6))    # (2 x 1) = 2 plots   |   shareX = True -> sharing same label for x-axis

ax[0].scatter(batters['avg'] , batters['strike_rate'] , color = 'red')            # 'ax[0]' is first 'axis object' of 'plt.subplots()' array
ax[1].scatter(batters['avg'] , batters['runs'])                   # 'ax[1]' is second 'axis object' of 'plt.subplots()' array

ax[0].set_title('Avg vs Strike Rate')
ax[1].set_title('Runs vs Avg')


ax[1].set_xlabel('Avg')       # here no need to set x-label for first graph i.e. ax[0] coz both are sharing same label i.e. 'Avg'

ax[0].set_ylabel('Strike Rate')
ax[1].set_ylabel('Avg')
```

```python
# 4 x 4 plot -> using 2-D array of axes-object having 4-axes

# fig and ax objects creation
fig , ax = plt.subplots(nrows = 2 , ncols = 2 , figsize = (10 , 5))            # nrows = 2 and ncols = 2


# creating plots using their valid axis object
ax[0 , 0].scatter(batters['avg'] , batters['strike_rate'] , color = 'red')     # ax[0 , 0] -> first object of 2-D array
ax[0 , 1].scatter(batters['avg'] , batters['runs'] , color = 'Blue')           # ax[0 , 1] -> second object of 2-D array
ax[1 , 0].hist(batters['avg'] , color = 'Green')                               # ax[1 , 0] -> third object of 2-D array
ax[1 , 1].hist(batters['runs'] , color = 'Orange')                             # ax[1 , 1] -> fourth object of 2-D array


fig.show()
```

# 3-D Surface Plot
* ### use in 'Loss Function' plottings in ML models

### meshgrid : makes all possible combinations without loop
- x = [1, 2, 3]   
-> 3-elements

- y = [10, 20]    
-> only 2-elements

- xx =
[[1  2  3]
[1  2  3]]    
-> row-wise repeated values

- yy =
[[10 10 10]
 [20 20 20]]  
 -> column-wise repeated values

```python
x = np.linspace(-10 , 10 , 100)
y = np.linspace(-10 , 10 , 100)

xx , yy = np.meshgrid(x , y)            # 'meshgrid' turns two lists 'x' and 'y' into all combinations 'xy' without loops where 'xx' -> repeated values row-wise and 'yy' -> repeated values column-wise

z = xx**2 + yy**2                       # creating 'z' variable with 'xx' and 'yy'

fig = plt.figure(figsize = (12 , 8))    # 'fig' object

ax = plt.subplot(projection = '3d')     # 'ax' object

p = ax.plot_surface(xx , yy , z , cmap = 'viridis')     # storing the plot in 'p' variable coz when we want 'colorbar' then we need to pass this variable into 'fig.colorbar(p)'
fig.colorbar(p)                                         # passing 'p' variable in 'fig.colorbar()'
```

# Contour Plots
* ### 3-D graph into 2-D

```python
z = xx**2 + yy**2                                  # 'xx' and 'yy' defined above

fig = plt.figure(figsize = (12 , 8))
ax = plt.subplot()

p = plt.contourf(xx , yy , z , cmap = 'viridis')   # 'contourf' -> contour fill
fig.colorbar(p)
```

# Heatmap - using 'plt.imshow(grid_data)' function
* ### use to compare two categorical variables side by side.
* ### grid data given to plot heatmap ( images are grid data hence images can be plot on 'heatmap' as well.

```python
# using 'IPL_ball_by_ball' dataset
delivery = pd.read_csv('/content/drive/MyDrive/CSV_files/Matplotlib Datasets/IPL_Ball_by_Ball_2008_2022.csv')
delivery.head()
```

```python
# Step-1- filter out those balls on which only '6' hitted
temp_df = delivery[(delivery['ballnumber'].isin([1 , 2 , 3 , 4 , 5 , 6])) & (delivery['batsman_run'] == 6)]
temp_df
```

```python
# Step-2- make a 'grid' by using pivot_table() fuunction
grid = temp_df.pivot_table(index = 'overs' , columns = 'ballnumber' , values = 'batsman_run' , aggfunc = 'count')
grid
```

```python
# Step-3- plotting the 'heatmap' using 'plt.imshow()'
plt.imshow(grid)                                               # function used to plot 'heatmap' in matplotlib
plt.xticks(np.arange(0 , 6) , list(range(1 , 7)))              # ranging ball number from 1-6 on x-axis
plt.yticks(delivery['overs'].unique() , list(range(1 , 21)))   # ranging overs from 1-20 on y-axis
plt.colorbar()
plt.show()

# y-axis showing over number
# x-axis showing ball number
```
