```python
# Speed of List and Numpy operations

# defining two lists 'a' and 'b'
a = [i for i in range(10000001)]             # 1 crore numbers
b = [i for i in range(10000000, 20000001)]   # 1 crore numbers

c = []    # new list but right now it is empty

import time
start = time.time()      # calculating the time at this moment before 'for' loop
for i in range(len(a)):
    c.append(a[i] + b[i])

time_list = time.time()- start
print(time_list)



# defining numpy array
import numpy as np
import time                  # importing 'time' module for calculating the time
a = np.arange(10000001)
b = np.arange(10000000, 20000001)

start = time.time()
c = a + b                    # adding two numpy arrays i.e. 'a' and 'b'
time_numpy = time.time() - start
print(time_numpy)



# checking that how bigger the time factor of list as compared to numpy
factor_of_difference = (time_list) / (time_numpy)
print(factor_of_difference)


'''
Note : why this time difference comes btw list and numpy operations?
           -List items are not directly stored at the memory but their addresses
           are stored and it is dynamic array which increases the size by 'double'
           everytime where as in 'Numpy array' the original value is itself stored
           at the memory which makes the operations faster than the 'List' and it
           is a static array which doesn't change it's size
'''
# Hence 'Numpy' is much more faster than the 'List'
```

# ***Advanced Indexing***

```python
# Normal indexing and slicing

a = np.arange(12).reshape(4,3)
print(a , '\n')

# fetching first-row
print(a[0])   # 0th row

# fetching the first-column
print(a[:, 0])    # ':' means all rows and 0th column

# finding 2-row and 3-column item
print(a[1,2] , '\n')    # indexes start from '0' for both rows and columns

# finding 2x2 matrix from 3-row and 1-column
print(a[2:4 , 0:2 ])    # 2:4 => 3 to 4 row , 0:2 => 1 to 2 column
# slicing includes 'start' index but excludes 'end' index here
```

```python
# providing indexes in the list
a = np.arange(24).reshape(6,4)
print(a , '\n')

# fetching 1, 3, 4 and 6 row by giving their indexes in the list
print(a[[0, 2, 3, 5]] , '\n')

# fetching 1, 3 and 4 column
print(a[:,[0, 2, 3]] , '\n')    # ':' means all rows and columnn list is [0, 2, 3]

# fetching pair of elements
print(a[[1,2] , [0,1]] , '\n')   # [1,2] => rows list and [0,1] => columns list
# Note : NumPy takes the first element from the row index list (which is 1) and the first element from the column index list (which is 0) and selects the element at a[1, 0].


# fetching cross-sectional matrix using 'np.ix_([rows_list] , [columns_list])
print(a[np.ix_([1,2],[0,1])])
```

# ***BroadCasting***
The term Broadcasting describes how NumPy treats arrays with different shapes during arithematic operations. Smaller array is 'broadcast' across the larger array so that they have compatible shapes

```python
# Same Shape Array operations

a = np.arange(6).reshape(2,3)
b = np.arange(6).reshape(2,3)

print(a , '\n')
print(b , '\n')
print(a + b)
```

## Broadcasting Rules
1-Make the two arrays have the same number of dimensions


*   if the number of dimensions of the two arrays are different, add new dimension with size '1' to the head of the array in the smaller dimension array.

2- Make each dimension of the two arrays, the same size


*   if the size of each dimension of the two arrays do not match, dimension with size '1' are stretched to the size of the other array.


*   if there is a dimension whose size is not '1' in either of the two arrays, it cannot be 'broadcasted', and an error is raised.

```python
# Mean Squared Error ( MSE ) => Used to calculate the error in the actual and predicted
#                               value from algorithm

#initializing 'actual' and 'predicted' values
actual = np.random.randint(1 , 50 , 25)        # random 25 numbers between 1-50
predicted = np.random.randint(1 , 50 , 25)     # random 25 numbers between 1-50

print(actual , '\n')      # actual array
print(predicted , '\n')   # predicted array

# function 'mse'
def mse( actual , predicted):
  return np.mean((actual - predicted)**2)    # 'mse' formula

# calling 'mse' function
print(mse(actual , predicted))
```

The Mean Squared Error (MSE) is a common metric used to evaluate the performance of regression models. It measures the average of the squared differences between the actual (observed) and predicted values.

The formula for MSE is:

$$ MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2 $$

Where:
- $ n $ is the number of data points.
- $ y_i $ is the actual value for the $ i $-th data point.
- $ \hat{y}_i $ is the predicted value for the $ i $-th data point.
- $ \sum $ represents the sum.

```python
# Binary cross entropy

import numpy as np

def binary_cross_entropy(y_actual, y_predicted):
  # Ensure predicted values are not exactly 0 or 1 to avoid log(0)
  epsilon = 1e-15
  y_predicted = np.clip(y_predicted, epsilon, 1 - epsilon)

  return -np.mean(y_actual * np.log(y_predicted) + (1 - y_actual) * np.log(1 - y_predicted))

# Example usage:
# Let's assume y_actual are the true labels (0s and 1s)
# and y_predicted are the model's predicted probabilities (between 0 and 1)

y_actual = np.array([1, 0, 1, 1, 0])
y_predicted = np.array([0.9, 0.1, 0.8, 0.6, 0.3])

loss = binary_cross_entropy(y_actual, y_predicted)
print(f"Binary Cross-Entropy Loss: {loss}")
```

Binary Cross-Entropy is a loss function used in binary classification problems. It measures the performance of a classification model whose output is a probability value between 0 and 1.

The formula for Binary Cross-Entropy is:

$$ BCE = -\frac{1}{n} \sum_{i=1}^{n} [y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i)] $$

Where:
- $ n $ is the number of data points.
- $ y_i $ is the actual label (0 or 1) for the $ i $-th data point.
- $ \hat{y}_i $ is the predicted probability (between 0 and 1) for the $ i $-th data point.
- $ \log $ is the natural logarithm.

```python
# np.append(array , number/array to be added) : it adds the values at the end of the numpy array

import numpy as np
a = np.arange(10)
print(a , '\n')
print(np.append(a , 10))                     # valued add at the end
print(np.append(a , [11 , 12 , 13]) , '\n')  # array of values added at the end

# appending in 2-D array
b = np.random.randint(1, 100 , 24).reshape(6,4)
print(b , '\n')
print(np.append(b , np.ones((b.shape[0] , 1)) , axis = 1))   # (b.shape[0] , 1) => giving the shape of the new array of 1's i.e. 6x1 ( this b.shape[0] is telling the shape of the rows in the array i.e. 6 )
print(np.append(b , np.ones((1 , b.shape[1])) , axis = 0))   # (1 , b.shape[1]) => giving the shape of the new array of 1's i.e. 1x6 ( this b.shape[1] gives the shape of the column in the array )

''' For a 2-dimensional array like b (which has a shape of 6 rows and 4 columns), b.shape will return the tuple (6, 4).
- b.shape[0] accesses the first element of this tuple, which is the number of rows. In your example, b.shape[0] is 6.
- b.shape[1] accesses the second element of this tuple, which is the number of columns. In your example, b.shape[1] is 4.

So, in the np.append() calls:
- np.ones((b.shape[0] , 1)) creates a new array of ones with the same number of rows as b (b.shape[0]) and 1 column i.e. '6x1'. This is done to create a column vector of ones that can be appended to the right side of 'b' (using axis=1).
- np.ones((1 , b.shape[1])) creates a new array of ones with 1 row and the same number of columns as b (b.shape[1]) i.e. '1x6'. This is done to create a row vector of ones that can be appended to the bottom of 'b' (using axis=0).
'''
```

## *Set Functions*

```python
# set numpy functions provide results in array form only

import numpy as np

m = np.array([1 , 2 , 3 , 4 , 5])
n = np.array([3 , 4 , 5 , 6 , 7])

# np.union1d()
print(f'Union : ' , np.union1d(m , n) , '\n')


# np.intersect1d()
print(f'Intersection : ' , np.intersect1d(m , n) , '\n')


# np.setdiff1d()
print(f'Setdiff : ' , np.setdiff1d(m , n) , '\n')   # m - n


# np.setxor1d()
print(f'XOR : ' , np.setxor1d(m , n) , '\n')  # XOR : elements in M and N both but not in common


# np.in1d()
print(np.in1d(m , 1) , '\n')

# Note : Use 'isin' method rather using 'in1d'
print(np.isin(m , 1))
```

### np.clip() : used to clip(limit) the values in an array : means all the values in an array greater than array_max will be equal to array_max and less than array_min will be converted to array_min and other remains same

```python
a = np.random.randint(1 , 100 , 20)
print(a , '\n')

print(np.clip(a , a_min = 25 , a_max = 75))
```
