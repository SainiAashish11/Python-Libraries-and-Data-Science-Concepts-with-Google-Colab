# Data Gathering from Web Scraping
* ### When we can't fetch from an API or API is not present, we use 'Web Scraping'

```python
# importing 'requests' and 'BeautifulSoup' library
import pandas as pd
import requests
from bs4 import BeautifulSoup   # library for web-scraping
```

```python
headers = {'User-Agent' : 'Mozilla/5.0 ( Windows NT 6.3; Win64; x64) AppleWebKit/537.36 ( KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
webpage = requests.get('https://books.toscrape.com/catalogue/page-1.html' , headers = headers).text
```

```python
# creating 'Soup' object of 'BeautifulSoup' library to use its functionalities
soup = BeautifulSoup(webpage , 'lxml')
```

```python
# to show the webpage's 'HTML' code in indented manner
# print(soup.prettify())
```

### Taking BooksToScrape Website to scrape the following data :
- Book_name
- Price
- Stock_Update

```python
# Step-1- To fetch out the Book_names from first_page
Books = soup.find_all('li' , class_ = 'col-xs-6 col-sm-4 col-md-3 col-lg-3')

Book_name = []                             # Empty list to append all the 'Book names'

for i in Books:                            # for-loop to fetch every 'Book' data present on 'first_page'
  # Find the 'h3' tag within the current 'li' item
  h3_tag = i.find('h3')
  # Find the 'a' tag within the 'h3' tag
  a_tag = h3_tag.find('a') if h3_tag else None

  # Check if an 'a' tag was found and if it has a 'title' attribute
  if a_tag and 'title' in a_tag.attrs:
    book_title = a_tag['title']            # storing the 'title' in 'book_title'
    Book_name.append(book_title)           # appending the 'book_title' in 'Book_name' list
    print(book_title)
  else:
    print("No title found for this book item.")
```

```python
# Step-2- Fetching all the prices of the books available on first page
Books = soup.find_all('li' , class_ = 'col-xs-6 col-sm-4 col-md-3 col-lg-3')

Price = []                                 # Empty list to append book prices
for i in Books:
  # Corrected to find a 'p' tag with class 'price_color'
  p_tag = i.find('p' , class_ = 'price_color')
  if p_tag:
    Price.append(p_tag.text.strip())       # Extract the text and append it to the Price list

print(Price)                               # Display the populated Price list
```

```python
# Step-3- Fetching stock update data
Books = soup.find_all('li' , class_ = 'col-xs-6 col-sm-4 col-md-3 col-lg-3')

Stock_update = []
for i in Books:
  p_tag = i.find('p' , class_ = 'instock availability')
  if p_tag:
    Stock_update.append(p_tag.text.strip())

print(Stock_update)
```

```python
pd.DataFrame({
    'Book_name' : Book_name,
    'Price' : Price,
    'Stock_check' : Stock_update
})
```

## Final DataFrame

```python
Book_data = pd.DataFrame()

for j in range(1, 51):
  url = 'https://books.toscrape.com/catalogue/page-{}.html'.format(j)
  headers = {'User-Agent' : 'Mozilla/5.0 ( Windows NT 6.3; Win64; x64) AppleWebKit/537.36 ( KHTML, like Gecko) Chrome/80.0.3987.162 Safari/537.36'}
  webpage = requests.get(url , headers = headers).text
  soup = BeautifulSoup(webpage , 'lxml')
  Books = soup.find_all('li' , class_ = 'col-xs-6 col-sm-4 col-md-3 col-lg-3')

  Book_name = []                             # Empty list to append all the 'Book names'
  Price = []                                 # Empty list to append book prices
  Stock_update = []

  # step-1
  for i in Books:                            # for-loop to fetch every 'Book' data present on 'first_page'
    # Find the 'h3' tag within the current 'li' item
    h3_tag = i.find('h3')
    # Find the 'a' tag within the 'h3' tag
    a_tag = h3_tag.find('a') if h3_tag else None

    # Check if an 'a' tag was found and if it has a 'title' attribute
    if a_tag and 'title' in a_tag.attrs:
      book_title = a_tag['title']            # storing the 'title' in 'book_title'
      Book_name.append(book_title)           # appending the 'book_title' in 'Book_name' list
    else:
      print("No title found for this book item.")

    # Step-2
    p_tag = i.find('p' , class_ = 'price_color')
    if p_tag:
      Price.append(p_tag.text.strip())       # Extract the text and append it to the Price list

    # Step-3
    p_tag = i.find('p' , class_ = 'instock availability')
    if p_tag:
      Stock_update.append(p_tag.text.strip())


  d = {
          'Book_name' : Book_name,
          'Price' : Price,
          'Stock_check' : Stock_update
          }

  df = pd.DataFrame(d)
  Book_data = pd.concat([df , Book_data] , ignore_index = True)
```
