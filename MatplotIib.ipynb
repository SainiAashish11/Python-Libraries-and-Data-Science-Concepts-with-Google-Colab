```python
# importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2-D plot
2-D plot.webp
- Always used for 'bivariate analysis' ( 2 columns )
- used for cat-cat , num-num or cat-num data-types

```python
# plotting a simple function
price = [48000 , 54000 , 57000 , 49000 , 47000 , 45000]    # first column
year = [2015 , 2016 , 2017 , 2018 , 2019 , 2020]           # second column

plt.plot(year , price)         # ( x-axis , y-axis )

plt.show()
```

```python
# from pandas DataFrame
batsmans = pd.read_csv('/content/drive/MyDrive/CSV_files/Matplotlib Datasets/sharma-kohli.csv')
display(batsmans)

plt.plot(batsmans['index'] , batsmans['V Kohli'])

plt.show()
```

```python
# plotting multiple plots
plt.plot(batsmans['index'] , batsmans['V Kohli'])      # blue-colour line
plt.plot(batsmans['index'] , batsmans['RG Sharma'])    # orange colour line

plt.show()
```

```python
# labels (axis names) and Title

plt.plot(batsmans['index'] , batsmans['V Kohli'])      # blue-colour line
plt.plot(batsmans['index'] , batsmans['RG Sharma'])

plt.title('Rohit Sharma vs Virat Kohli Runs Comparisons over the years')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.show()
```

```python
# colors(name or hex code)
plt.plot(batsmans['index'] , batsmans['V Kohli'] , color = 'black' )      # color = 'black'
plt.plot(batsmans['index'] , batsmans['RG Sharma'], color = '#C1E00B')    # color = 'hex-code'


plt.title('Rohit Sharma vs Virat Kohli Runs Comparisons over the years')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.show()
```

```python
# line ( style and width )
# linestyles -> solid , dashed , dotted , dashdot
# linewidth -> any number

plt.plot(batsmans['index'] , batsmans['V Kohli'] , color = 'black' , linestyle = 'solid' , linewidth = 3)          # linestyle = 'solid' and linewidth = 3
plt.plot(batsmans['index'] , batsmans['RG Sharma'], color = '#C1E00B' , linestyle = 'dashdot' , linewidth = 2)     # linestyle = 'dashdot' and linewidth = 2


plt.title('Rohit Sharma vs Virat Kohli Runs Comparisons over the years')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.show()
```

```python
# marker -> shows the connecting points of 'x-axis' and 'y-axis'
# type and size
plt.plot(batsmans['index'] , batsmans['V Kohli'] , color = 'black' , linestyle = 'solid' , linewidth = 3 , marker = '*' , markersize = 10)           # marker = '*' and markersize = 10
plt.plot(batsmans['index'] , batsmans['RG Sharma'], color = '#C1E00B' , linestyle = 'dashdot' , linewidth = 2 , marker = '^' , markersize = 8)       # marker = '^' and markersize = 8


plt.title('Rohit Sharma vs Virat Kohli Runs Comparisons over the years')
plt.xlabel('Season')
plt.ylabel('Runs Scored')
plt.show()
```

```python
# legend() -> used to show the labels of the plot
# legend(loc = '') : locations of the lable -> best , upper right , upper left , lower left , lower right , right , center left , center right , lower center
# label : tagging the plot with specific name

plt.plot(batsmans['index'] , batsmans['V Kohli'] , color = 'black' , linestyle = 'solid' , linewidth = 3 , marker = '*' , markersize = 10 , label = 'Virat')         # label = 'Virat'
plt.plot(batsmans['index'] , batsmans['RG Sharma'], color = '#C1E00B' , linestyle = 'dashdot' , linewidth = 2 , marker = '^' , markersize = 8 , label = 'Rohit')     # label = 'Rohit'


plt.title('Rohit Sharma vs Virat Kohli Runs Comparisons over the years')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend(loc = 'best')
plt.show()
```

```python
# limiting axes -> fixing the range of the data values on the axes
# Note : we do this when outlier value comes in our data and data pattern vanishes

price = [48000 , 54000 , 57000 , 49000 , 47000 , 4500000]    # price - first column
year = [2015 , 2016 , 2017 , 2018 , 2019 , 2020]             # year  -  second column

plt.plot(year , price)

plt.ylim(0 , 100000)     # y-axis limit
plt.xlim(2017 , 2020)    # x-axis limit
plt.show()
```

```python
# grid() : increases the readibility of the graph using horizontal and vertical lines

plt.plot(batsmans['index'] , batsmans['V Kohli'] , color = 'black' , linestyle = 'solid' , linewidth = 3 , marker = '*' , markersize = 10 , label = 'Virat')
plt.plot(batsmans['index'] , batsmans['RG Sharma'], color = '#C1E00B' , linestyle = 'dashdot' , linewidth = 2 , marker = '^' , markersize = 8 , label = 'Rohit')


plt.title('Rohit Sharma vs Virat Kohli Runs Comparisons over the years')
plt.xlabel('Season')
plt.ylabel('Runs Scored')

plt.legend(loc = 'best')

plt.grid()            # grid() function for horizon
plt.show()
```

```python
# show() : show the graps on any editor while using matplotlib
price = [48000 , 54000 , 57000 , 49000 , 47000 , 45000]    # price - first column
year = [2015 , 2016 , 2017 , 2018 , 2019 , 2020]             # year  -  second column

plt.plot(year , price)

plt.show()
```

## Scatter Plots
scatter plot.webp

- used for Bivariate analysis
- best used in numerical vs numerical data
- Use Case : Finding Correlation between two quantities

```python
# plt.scatter() function
x = np.linspace(-10 , 10 , 50)
print(x , '\n')

y = 10*x + 3 + np.random.randint(0 , 300 , 50)
print(y , '\n')

plt.scatter(x , y)

plt.show()
```

```python
# plt.scatter on pandas data
df = pd.read_csv('/content/drive/MyDrive/CSV_files/Matplotlib Datasets/batter.csv')
display(df)

# fetching first 50 batsmans and storing in the main DataFrame
df = df.head(50)

# visualize scatter plot of 'avg' and 'strike_rate'
plt.scatter(df['avg'] , df['strike_rate'])
plt.title('Avg and SR analysis of first 50 batsmans')
plt.xlabel('Average')
plt.ylabel('SR')
```

```python
# size parameter - s

# initialize the 'tips' DataFrame from seaborn library
tips = sns.load_dataset('tips')

plt.scatter(tips['total_bill'] , tips['tip'] , s = tips['size']*20)    # s = tips['size'] -> is for changing the size of the dot w.r.t. the number of customers brought with them. Multiplied by '20' to just show the enlarged dots clearly

plt.show()
```

```python
# scatter-plot using plt.plot
# faster technique as compared to plt.scatter() when large data is present( crores of data points )

plt.plot(tips['total_bill'] , tips['tip'] , 'o')        # 'o' is used to change the line plot as scattered dots

plt.show()
```

# Bar Chart
bar chart.webp

## Two types of Bar charts :
- Vertical
- Horizontal

- Good for Bivariate analysis
- Generally used for num-cat data-type
- Use Case : Aggregate analysis of groups

```python
# simple bar chart ( default - vertical )
children = [10 , 20 , 40 , 10 , 30]
colors = ['red' , 'blue' , 'green' , 'yellow' , 'pink']

plt.bar(colors , children)

plt.show()
```

```python
# plt.barh() : Horizontal bar chart
children = [10 , 20 , 40 , 10 , 30]
colors = ['red' , 'blue' , 'green' , 'yellow' , 'pink']

plt.barh(colors , children)

plt.show()
```

```python
# colors and label

children = [10 , 20 , 40 , 10 , 30]
colors = ['red' , 'blue' , 'green' , 'yellow' , 'pink']

# color
plt.barh(colors , children , color = 'purple')    # color = 'purple'

# title and labels
plt.title('Children and their color choices')
plt.xlabel('Color')
plt.ylabel('Number of children')
plt.show()
```

```python
# multiple bar charts
df = pd.read_csv('/content/drive/MyDrive/CSV_files/Matplotlib Datasets/batsman_season_record.csv')
df
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2 , df['2015'] , width = 0.2 , color = 'yellow')    # np.arange(df.shape[0]) - 0.2 -> df.shape[0] = 5 , means 0.2 is subtracting from range of number from 0-4 to shift the bar plot at left side
plt.bar(np.arange(df.shape[0]) , df['2016'] , width = 0.2 , color = 'red')
plt.bar(np.arange(df.shape[0]) + 0.2 , df['2017'] , width = 0.2 , color = 'blue')      # shifting bar plot to right side by 0.2 ( Note : 0.2 is just hit and trial number here )

plt.xticks(np.arange(df.shape[0]) , df['batsman'])    # to replace the 'numbers' with 'batsman names' on x-axis

plt.show()
```

```python
# a problem -> categories names overlapping
children = [10 , 20 , 40 , 10 , 30]
colors = ['red red red red' , 'blue blue blue blue' , 'green green green green' , 'yellow yellow yellow yellow' , 'pink pink pink pink']

plt.bar(colors , children , color = 'purple')

plt.xticks(rotation = 'vertical')          # rotating the color names to vertical
plt.show()
# Note : plt.yticks('horizontal'/'vertical') -> for y-axis values
```

```python
# stacked bar chart
plt.bar(df['batsman'] , df['2017'] , label = '2017')
plt.bar(df['batsman'] , df['2016'] , bottom = df['2017'] , label = '2016')                 # bottom = df['2017'] -> means 2016 ka plot hame 2017 ke upar chahiye
plt.bar(df['batsman'] , df['2015'] , bottom = df['2016'] + df['2017'] , label = '2015')    # bottom = df['2016'] + df['2017'] -> means 2015 ka plot hame 2016 and 2017 ke sum ke upar chahiye

plt.legend(loc = 'best')
plt.show()
```

## Histogram
- used to visualize the data into distinct groups/intervals which we say 'bins'

Histogram.webp

- Univariate analysis
- Numerical column
- Use case : Frequnecy Count

```python
# simple histogram plot for numerical data
data = [32 , 45 , 56 , 10 , 15 , 27 , 61]

plt.hist(data , bins = [10 , 25 , 40 , 55 , 70])

plt.show()
```

### logarithmic scale
- Use to make the valid data bins so that mostly data shouldn't fit into few bins only

```python
arr = np.load('/content/drive/MyDrive/CSV_files/Matplotlib Datasets/big-array.npy')    # loading a big numpy array in 'arr' variable
print(arr)

# plotting the histogram with normal scale
plt.hist(arr)                                    # data is saturated in 30-40 bin only

plt.show()
plt.xlabel('Number')
plt.ylabel('Repetition of Number')
```

```python
# plotting the histogram using logarithmic scale
plt.hist(arr , log = True)

plt.show()
plt.xlabel('Number')
plt.ylabel('Repetition of Number')
```

# Pie Chart
- Use to plot on different categories and tell their contribution in total value
pie chart.webp

- Univariate/Bivariate analysis
- Categorical vs Numerical
- Use case : To find contribution on a standard scale

```python
# simple data
data = [10 , 20 , 33 , 12 , 32]
subjects = ['eng' , 'science' , 'maths' , 'sst' , 'hindi']

plt.pie(data , labels = subjects)         # labels = subjects -> classifies the subjects on the pie chart
plt.show()
```

```python
# pie chart on real dataset

df = pd.read_csv('/content/drive/MyDrive/CSV_files/Matplotlib Datasets/gayle-175.csv')
df
```

```python
plt.pie(df['batsman_runs'] , labels = df['batsman'])

plt.show()
```

```python
# percentage fpr each category on pie chart
plt.pie(df['batsman_runs'] , labels = df['batsman'] , autopct = '%0.1f%%')

plt.show()
```

```python
# colors
plt.pie(df['batsman_runs'] , labels = df['batsman'] , autopct = '%0.1f%%' , colors = ['blue' , 'green' , 'yellow' , 'pink' , 'cyan' , 'brown'])

plt.show()
```

```python
# explode : to focus out any specific category/data from the pie chart
plt.pie(df['batsman_runs'] , labels = df['batsman'] , autopct = '%0.1f%%' , explode = [0.3 , 0 , 0 , 0 , 0 , 0])

plt.show()
```

```python
 # shadow : shadowing the pie chart
plt.pie(df['batsman_runs'] , labels = df['batsman'] , autopct = '%0.1f%%' , explode = [0.3 , 0 , 0 , 0 , 0 , 0] , shadow = True)

plt.show()
```
