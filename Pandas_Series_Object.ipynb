# Making of Series object using Python List datatype

```python
# String series
country = ["India" , "Pakistan" , "USA" , "Nepal" , "SriLanka"]
pd.Series(country)   # Series() function makes any list into column data

''' Series object consists of:
1- Index : serial number assigned to data-value ( automatically generated )
2- Value : data-value after assigned index ( Note : data-type of 'value' is object)

Note : 99% of times Pandas consider 'string' data type as Object but not everytime
'''
```

```python
# integers series
runs = [31 , 24 , 56 , 78 , 100]
pd.Series(runs)
```

```python
# Custom Index writing
marks = [67 , 57 , 89 , 100]
subject = ['maths' , 'english' , 'science' , 'hindi']

pd.Series(marks, index = subject)
```

```python
# setting a 'name' to a series
pd.Series(marks, index = subject , name = 'Aashish ke marks')
```

# Series Methods

```python
# head(no. of values) : gives the number of mentioned values (default = 5 rows) from start

print(my_series.head() , '\n')   # first 5 data-values | 'my_series' is the variable in which we stored 'Series' in above code cell
print(my_series.head(10), '\n')  # 10 data-values from the beginning

# tail(no. of values) : gives the values from the end (default = 5 rows)
print(my_series.tail() ,'\n')   # last 5 data-values
print(my_series.tail(10) , '\n')  # last 10 data-values
```

## Series Indexing

```python
# integer indexing
x = pd.Series([12 , 13 , 1 , 4 , 5 , 7 , 10 , 100 , 78 , 67])
print(x[1])
```

```python
print(marks_series , '\n')

# .iloc[] : position based indexing is done with this. Positive and Negative indexing
# negative indexing : doesn't work in integer based indexing directly and we have to use '.iloc[]'
print(x.iloc[-1] , '\n')

# negative indexing for 'string' based indexes works normally
print(marks_series.iloc[-1])   # Use .iloc[] for position-based negative indexing
```

```python
# positive slicing : multiple items fetching
print(marks_series.iloc[1:5] , '\n')
print(marks_series.iloc[3:6] , '\n')

# jump slicing
print(marks_series.iloc[1:5:2] , '\n')  # second element from the current element
print(marks_series.iloc[::2])           # fetching alternate data-values

# negative slicing
print(marks_series.iloc[-5:-1] , '\n')   # -5 to second last ( -1 is excluded )
print(marks_series.iloc[-5:] , '\n')     # -5 to last element
print(marks_series.iloc[-5::2])          # jumping of 2 from -5 index to last element
```

```python
# negative jumping

print(marks_series , '\n')     # original 'marks_series' data

print(marks_series.iloc[::-2] , '\n')  # jump of -2
print(marks_series.iloc[-2:0:-3])      # jumo of -3
```

```python
# fancy indexing
print(marks_series.iloc[[1 , 3 , 4 , 6]])     # list of indexes to be fetched
```

```python
# indexing with 'labels'
print(marks_series.loc['English': 'Economics'] , '\n')  # 'label' base slicing
print(marks_series[['Hindi' , 'Maths' , 'CS']])         # fancy indexing with 'labels'
```

```python
# astype : changes the data type of the data-value
import sys
print(sys.getsizeof((marks_series.astype(np.int16) )) , '\n')           # int64 -> int16

# printng the 'size of marks_series' after converting data type from 'int64' -> 'int16'
print(sys.getsizeof(marks_series) , '\n')

# 'integer' -> 'float' conversion
print(sys.getsizeof((marks_series.astype(np.float64))))
```

```python
# between : return boolean result. It tells whether something exists in a range or not
marks_series[marks_series.between(50 , 90)]    # 50 and 90 are included
```

```python
# clip : values become 'min' and 'max' as per the given values as argument
marks_series.clip(75 , 90)      # data-values less than 75 will convert to 75 and larger than 90 will become 90, and others remain as it is
```

```python
# drop_duplicates() : drops the duplicate values and keeps first-occurence(default) only
temp = pd.Series([1 , 1 , 2 , 2 , 3 , 4 , 6 , 6 , 7 , 7 , 7 , 8 , 10])      # new series formed just now
print(temp , '\n')
print(temp.drop_duplicates() , '\n')

# drops the first occurence
print(temp.drop_duplicates(keep = "last") , '\n')



# duplicated() : tell the duplicate values present in the data-values
print(temp.duplicated() , '\n')

# how many duplicated values in the data
print(temp.duplicated().sum())    # sum() will add all True( i.e. 1) values
```

```python
# isnull : how many missing values ( NaN )
temp = pd.Series([1 , 2 , 3 , np.nan , 5 , np.nan , 6 , 7 , np.nan])   # new series

print(temp.isnull().sum())    # isnull() gives the boolean series having indexes True/False and sum() adds the 'True' values
```

```python
# dropna : drops the missing values and gives new series with genuine values
temp.dropna()
```

```python
# fillna : fill the missing values
print(temp.fillna(0) , '\n')      # replacing missing values with '0'

print(temp.fillna(temp.mean()))   # replacing missing values with 'mean' of the data
```

```python
# isin([]) : values are in data or not
print(temp.isin([4 , 6 , 7]) , '\n')    # boolean series

temp[temp.isin([4 , 6 , 7])]            # boolean masking
```

```python
# apply() : helps to apply custom logic on series

# apply() with 'lambda' function
print(marks_series.apply(lambda x: 'Good marks' if x > marks_series.mean() else 'Bad marks' ) , '\n')   # if marks are greater than 'mean' marks then 'Good marks' otherwise 'Bad marks'


# apply() with function passing
def grade(x):
  if x > marks_series.mean():
    return 'Good marks'
  else:
    return 'Bad Marks'

marks_series.apply(grade)    # grade() function passed into apply() method and data-values from marks_series are accessing one by one by grade() function
```
