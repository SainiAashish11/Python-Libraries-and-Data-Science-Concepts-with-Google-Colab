```python
import seaborn as sns
import plotly.express as px
```

```python
# importing datasets
tips = sns.load_dataset('tips')
iris = sns.load_dataset('iris')
```

# 1- Categorical Scatter plots
## 1.1- Strip Plot

### Axes level function

```python
# using axes-level function -> sns.stripplot()
sns.stripplot(data = tips , x = 'day' , y = 'total_bill')
```

```python
# 'jitter' parameter -> adds th noise in the data values so that visibility should be good, if jitter = False then data points will come in one-single line
sns.stripplot(data = tips , x = 'day' , y = 'total_bill' , jitter = False )
```

### Figure level function
- In every 'Figure-level' function -> Faceting can be done using 'col' and 'row' parameter

```python
# using figure-level function -> sns.catplot()
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip')
```

```python
# 'jitter' value can be increased or decreased with values
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip' , jitter = 0.02)
```

```python
# 'hue' parameter
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip' , jitter = 0.2 , hue = 'sex')
```

```python
# faceting
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'strip' , jitter = 0.2 , col = 'smoker' , row = 'sex')
```

### 1.2- Swarm Plot - Represents the data values more originally distributed but good for small and medium size datasets otherwise it spread too out and collapse in each other like tree branches

```python
 # using axes-level function -> sns.swarmplot()
sns.swarmplot(data = tips , x = 'day' , y = 'total_bill' , hue = 'sex')
```

```python
# using figure-level function -> sns.catplot()
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'swarm' , hue = 'sex')
```

## 2- Categorical Distribution Plots

### 2.1- Box Plot - It is an standardized way of displaying the distribution of data based on a five number summary( 'minimum' , first-quartile[Q1] , 'median' , 'third-quartile[Q3]' and 'Maximum'. It can tell you about your outliers and what their values are. Boxplots can also tell you if your data is symmetrical , how tightly grouped and if and how your data is skewed.
image.png

```python
# using axes-level function -> sns.boxplot()
sns.boxplot(data = tips , x = 'sex' , y = 'total_bill')
```

```python
# using figure-level function -> sns.catplot()
sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , kind = 'box')    # male/female total_bill = 2 plots
```

```python
# 'hue' parameter
sns.boxplot(data = tips , x = 'day' , y = 'total_bill' , hue = 'sex')   # every day male/female total_bill spending box plots = 8 plots
```

```python
# single box plot -> numerical column
sns.boxplot(data = tips , y = 'total_bill')
```

### 2.2- Violin Plot = Box Plot + Kde Plot

```python
# using axes-level function -> sns/violinplot()
sns.violinplot(data = tips , x = 'day' , y = 'total_bill')
```

```python
# using figure-level function -> sns.catplot()
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'violin')
```

```python
# 'hue' parameter
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'violin' , hue = 'sex')
```

```python
# 'split' parameter -> used to split the plots on the basis of categories in the 'hue' parameter
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'violin' , hue = 'sex' , split = True)
```

```python
# faceting using 'col' and 'row' parameters
sns.catplot(data = tips , x = 'day' , y = 'total_bill' , kind = 'violin' , col = 'smoker' , row = 'sex')
```

## 3- Categorical Estimate Plot
### 3.1- Bar Plot -> Calculating the mathematical estimate of categorical data with numerical data ( mean , mdeian , max , min ,  std deviation , variance , etc )


```python
# using axes-level function -> sns.barplot()
sns.barplot(data = tips , x = 'sex' , y = 'total_bill')
```

```python
# using figure-level function -> sns.catplot()
sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , kind = 'bar')
```

### When there are mutliple observations in each categorical column ( sex -> male/female ), 'bar plot' uses bootstrapping to compute a 'confidence interval' arond the estimate, which is plotted using 'error bars'

```python
# 'hue' parameter -> color grading
sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , kind = 'bar' , hue = 'smoker')
```

```python
# 'estimator' parameter -> calculates the mathematical function ( default = mean )
import numpy as np

sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , kind = 'bar' , hue = 'smoker' , estimator = np.std)    # np.std -> 'standard deviation' using numpy library
```

```python
# 'ci' parameter -> confidence interval ( error bars )
import numpy as np

sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , kind = 'bar' , hue = 'smoker' , estimator = np.std , errorbar = None)    # np.std -> 'standard deviation' using numpy library
```

### 3.2- Point Plot -> Focus on difference between the plots
- Connects the highest points of the bar and removes the error bars

```python
# using axes-level function -> sns.pointplot()
sns.pointplot(data = tips , x = 'sex' , y = 'total_bill' , hue = 'smoker' , estimator = 'min')                # estimator = 'min'
```

```python
# using figure-level function -> sns.catplot()
sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , kind = 'point' , hue = 'smoker' , estimator = 'var' , errorbar = None)   # estimator = var -> variance
```

### 3.3- Count Plot - Counts the overall categories and shows a quantitative data
- only 'one' data column is allowed for analysis

```python
# using axes-level function -> sns.countplot()
sns.countplot(data = tips , x = 'sex')
```

```python
# 'hue' parameter
sns.countplot(data = tips , x = 'sex' , hue = 'day')         # how many male and female people came on different days( thur , fri , sat , sun )
```

```python
# using figure-level function -> sns.catplot()
sns.catplot(data = tips , x = 'sex' , hue = 'day' , kind = 'count')
```

### faceting using 'catplot()'

```python

sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , col = 'smoker' , row = 'time' , kind = 'box')
sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , col = 'smoker' , row = 'time' , kind = 'point')
sns.catplot(data = tips , x = 'sex' , y = 'total_bill' , col = 'smoker' , row = 'time' , kind = 'violin')
```

# Regression Plots
* ### Displays the relationship between two variables, typically a 'dependent variable' and an 'independent variable.'
* ### It plots the scatter plot of the data points and then create a regression line through the data points
* ### Also creates a 95% confidence interval of that regression

## Types of regressions plots
### 1- regplot

### 2- Implot

```python
# regplot -> axes-level function
# hue parameter is not available in it
sns.regplot(data = tips , x = 'total_bill' , y = 'tip')

# Note : shaded area near line is showing error interval
```

```python
# lmplot -> figure-level function
# 'hue' parameter is available in it
sns.lmplot(data = tips , x = 'total_bill' , y = 'tip' , hue = 'sex')    # 2 colored plots present using 'hue' parameter
```

```python
# residplot : plots the error of the data point w.r.t the regression line. Data points above 'zero-reference' line shows the error of that data point and point on the line shows 'no-error'
sns.residplot(data = tips , x = 'total_bill' , y = 'tip')
```

# 6- Multi-grid Plots :
* ### 1- Facet Grid -> A second way to plot Facet plots. Provides much more options to customize the plots
* ### 2- Pair Plots/Grid -> Plot pairwise relationship between the numerical columns of the dataset
* ### 3- Joint Plots/Grid -> Creates 'histogram' on individual numerical columns wiht the demanded(jo tumne banaya hai) plot

### 6.1- Facet Plot

```python
g = sns.FacetGrid(data = tips , col = 'day' , row = 'time')
g.map(sns.violinplot , 'sex' , 'total_bill', order=tips['sex'].unique())   # 'order' parameter tell the Facet Grid to plot 'x-axis' category in the following order
```

```python
#  g.add_legend() -> to provide the info of labelling
g = sns.FacetGrid(data = tips , col = 'day' , row = 'time' , hue = 'smoker')   # making facet grid object
g.map(sns.violinplot , 'sex' , 'total_bill', order=tips['sex'].unique())       # details of the plot that we needed
g.add_legend()                                                                 # use to show legend box
```

## 6.2- Pair Plot vs pair Grid
- 1- Pair Plot -> Plots the Pairwise plots on the numerical data
- 2- Pair Grid -> Plots the Pairwise plots on the numerical data but provides more customization like 'Facet Grid' as compared to 'Facet Plots'

### 6.2.1- Pair Plot

```python
# using 'iris' flower wala dataset
sns.pairplot(data = iris)   # 4x4 = 16 plots -> '4 columns on x-axis' and same '4 columns on y-axis'

# Note : where x-axis and y-axis columns are same -> plot is 'histogram'
```

```python
# 'hue' parameter
sns.pairplot(data = iris , hue = 'species')

# Note : for same columns -> 'kde' plot is plotted
```

## 6.2.2- Pair Grid

```python
g = sns.PairGrid(data = iris , hue = 'species')

# g.map() to define the plots
g.map(sns.scatterplot)
```

```python
# map_diag() -> diagonal plots
# map_offdiag() -> other diagonal plots
g = sns.PairGrid(data = iris , hue = 'species')   # 'g' -> pair grid object

# g.map() to define the plots using 'g' object
g.map_diag(sns.boxplot)
g.map_offdiag(sns.scatterplot)
```

```python
# map_diag -> map_upper() and map_lower()
g = sns.PairGrid(data = iris , hue = 'species')

g.map_diag(sns.histplot)         # diagonal plots
g.map_upper(sns.kdeplot)         # upper-diagonal plots
g.map_lower(sns.scatterplot)     # lower-diagonal plots
```

```python
# vars[] -> from which columns i want pairs of plots
g = sns.PairGrid(data = iris , hue = 'species' , vars = ['sepal_width' , 'petal_width'] )

g.map_diag(sns.histplot)
g.map_upper(sns.kdeplot)
g.map_lower(sns.scatterplot)
```

## 6.3- Joint Plots vs Joint Grid
* ### use to analyse 1-D and 2-D plots at a time

### 6.3- Joint Plot

```python
sns.jointplot(data = tips , x = 'total_bill' , y = 'tip' , kind = 'scatter' , hue = 'sex')

# Note : kind = ['scatter' , 'hist' , 'hex' , 'kde' , 'reg' , 'resid'] -> yahi 'kind' parameter mein diye jaa sakte hai
# Note : Use of `hue` with `kind=['hex' , 'reg' , 'resid']` is not currently supported
# Note : Plotted '1-D kde' plot on individual columns but '2-D kde' plot on both columns -> so joint plots used for 1-D and 2-D analysis at a time
```

### 6.3- Joint Grid

```python
g = sns.JointGrid(data = tips , x = 'total_bill' , y = 'tip')   # 'g' is 'Jointgrid' Object

g.plot(sns.scatterplot , sns.kdeplot)                          # 'scatterplot' for 2-D analysis | 'histplot' for 1-D analysis on individual columns

# Note : Here you can plot any type of plot i.e. not limited to Joint Plot's 'kind' parameter list of supported plots
```
