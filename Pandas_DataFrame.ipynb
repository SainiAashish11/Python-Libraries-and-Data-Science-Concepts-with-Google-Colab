### drop_duplicates()

```python
# drop_duplicates( series + DataFrame ) : removes the duplicate 'rows in DataFrame'
students_marks = pd.DataFrame({
    'Subjects' : ['Hindi' , 'English' , 'CS' , 'English'],
    'Marks'  : [90 , 89 , 98 , 89]
})

print(students_marks , '\n')    # displaying 'students_marks' data

print(students_marks.drop_duplicates() , '\n')   # Note : removes the duplicate row ( by default removes the last occurence )

# keeps the last occurence
students_marks.drop_duplicates(keep = 'last')
```

```python
# Question : find the last match played by Virat Kohli in Delhi

# Step:1 - Make a new list containg all players from 'Team1' and 'Team2'
ipl['all_players'] = ipl['Team1Players'] + ipl['Team2Players']
ipl

# Step:2 - now check, whether 'V Kohli' is in the 'all_players' list or not

# function to find 'V Kohli' in the 'all_players' list or not
def did_kohli_play(players_list : list) -> bool:
  return 'V Kohli' in players_list

ipl['kohli_played'] = ipl['all_players'].apply(did_kohli_play)    # taking data into new column 'Kohli_played'


# Step:3 - now find the data where 'City' == 'Delhi' and 'did_kohli_play' == True and save it in updated 'ipl' DataFrame
ipl = ipl[(ipl['City'] == 'Delhi') & (ipl['kohli_played'] == True)]

# Step:3 - drop the rows on the basis of 'City' and 'kohli_played' columns so that only one latest row can come
ipl.drop_duplicates(subset = ['City' , 'kohli_played'])

```

### apply()

```python
# apply(series + DataFrame) : add customized function to any method present

''' apply() on series data'''

# making a new series 'temp'
temp = pd.Series([10 , 98 , 22 , 77 , 100])
print(temp)

# sigmoid function
def sigmoid(value : int ) -> float:
  return 1/1 + np.exp(-value)

# passing 'sigmoid' function into 'apply()
temp.apply(sigmoid)
```

```python
''' apply() on DataFrame '''

# creating new Dataframe
points_df = pd.DataFrame({
    '1st_point' : [(3,4) , (-6,5) , (0,0) , (-10,1) , (4,5)],
    '2nd_point' : [(-3,4) , (0,0) , (2,2) , (10,10) , (1,1)]
})

# printing DataFrame
print(points_df , '\n')

# euclidean_dist() function
def euclidean_dist(row):
  pt_A = row['1st_point']
  pt_B = row['2nd_point']
  return ((pt_A[0] - pt_B[0])**2 + (pt_A[1] + pt_B[1])**2)**0.5

# passing euclidean_dist() function to 'apply()

points_df['Distance'] = points_df.apply(euclidean_dist , axis = 1)   # 'axis = 1' means we are sending rows to 'euclidean_dist()' function
points_df
```
