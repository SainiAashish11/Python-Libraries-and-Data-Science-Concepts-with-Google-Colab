```python
import numpy as np
import pandas as pd
```

## What are Vectorized operations
- Vectorized operations = fast, clean, NumPy-backed operations that work on entire pandas Series/DataFrames at once, without Python loops.

```python
# Example of vectorized operation

a = np.array([1 , 2 , 3 , 4])
display(a)

display(a * 4)   # every element of array 'a' is multiplied by '4' coz array 'a' taken as one vector
```

### Problem in vectorized operations in vanilla python ( basic )
- Sometimes if any other Data-type is present inside 'list' , 'set' , 'tuple' or 'dictionary' then python's in-built operations will failed to solve the problem

```python
# Example of failing for vanilla python
s = ['cat' , 'mat' , None , 'rat']

[i.startswith('c') for i in s]           # here 'None' data-type will not be accessed by 'startswith()' function
# AttributeError: 'NoneType' object has no attribute 'startswith'
```

### How Pandas solves the problem

- In pandas, the string accessor .str is a specialized interface that allows you to perform vectorized string operations on an entire Series or Index containing text data.

### 1- Why '.str' Is Needed
- A pandas Series is not a plain list of Python strings; therefore, you cannot directly call .lower(), .split(), etc., on the Series.

Ex ⚡

- s = pd.Series(["Aashish", "ENGINEERING", "india"])

  s.lower()     # ERROR

Solution ✅
- s.str.lower()

### 2- How '.str' Works
- '.str' enables vectorized string operations across the entire Series, meaning the operation will run on every element automatically.


```python
s = pd.Series(['cat' , 'mat' , None , 'rat'])
display(s)

# using string accessor 'str' and 'startswith()' function
s.str.startswith('c')

# Note : This operation is fast and optimized due to 'pandas operations' which built on 'numpy' which built on 'C-language'
```

## Working on real world Dataset 'titanic_csv'

```python
df = pd.read_csv('/content/drive/MyDrive/CSV_files/titanic.csv')
df.head()
```

```python
# extract 'Name' column only as Series
Names = df['Name']
Names
```

### Common functions on Strings

```python
# lower() : all alphabets of string will become small
Names.str.lower()
```

```python
# upper() : all string aphabets capital
Names.str.upper()
```

```python
# capitalize() : only first alphabet of whole string will be capital
Names.str.capitalize()
```

```python
# title() : every word ka first alphabet will be capital
Names.str.title()
```

```python
# len() : number of alphabets/characters(including -> spaces) in string
Names.str.len()
```

```python
# strip() : removes leading(in front)/trailing(at end) spaces
Names.str.strip()

# useful function in 'NLP' related tasks
```

```python
# split() : splits the string on given 'character'
# get() : fetches particular list indexes after splitting

tukde = Names.str.split(',')     # splitting on basis of comma -> ','
display(tukde)

first_name = tukde.str.get(0)    # fetching '0th' item -> 'first_name' from the list of splitted words
display(first_name)


# storing 'first_name' in main dataset
df['first_name'] = first_name
```

```python
# split and control using 'n' parameter which limits the splitting until given number
display(Names.str.split(',').str.get(1).str.strip().str.split(' ' , n = 1))


# storing 'title' and 'endname' in main dataset now
df[['title' , 'endname']] = Names.str.split(',').str.get(1).str.strip().str.split(' ' , n = 1 , expand = True)               # 'expand = True' for converting 'series' to a 'DataFrame'
df.head()                                               # 'first_name' , 'title' and 'endname' columns are present in the main data now
```

```python
# replace() : replaces the given string with other string

df['title'] = df['title'].str.replace('Ms.' , 'Miss.')        # 'Ms' is replaced by 'Miss' in 'title' column and changes saved in main dataset
# df['title'] column updated
```

## Filtering

```python
# startswith() : string started with given character/alphabet
df[df['first_name'].str.startswith('A')]
```

```python
# endswith() : ending with given character/alphabet
df[df['endname'].str.endswith('y')]
```

```python
# isdigit() : returns 'True' if string having all characters are digits
df['Name'].str.isdigit()
```

```python
# isalpha() : returns 'True' if all string characters are 'alphabets' only
df['first_name'].str.isalpha()
```

## Regex ( regular expression ) using contains() function
- used for advanced filtering

```python
# search John in Names in both upper or lower cases
df[Names.str.contains('John' , case = False)]
```

```python
# find lastnames starting and ending with an vowel
df[df['endname'].str.contains(r'^[aeiouAEIOU].+[aeiouAEIOU]$')]

''' Explanation :
1-  r'...' → raw string to avoid escaping issues.
2-  ^[aeiouAEIOU] → starts with vowel.
3-  .+ → has one or more characters in between.
4-  [aeiouAEIOU]$ → ends with vowel.
'''
```

```python
# find firstnames strating and ending with consonants
df[df['first_name'].str.contains(r'^[^aeiouAEIOU].+[^aeiouAEIOU]$')]


''' [^aeiouAEIOU] -> '^' inside the list means the complement of aeiouAEIOU which is a 'consonant'
```
